```python
import pandas as pd

df = pd.read_csv("support_tickets.csv")

print("Dataset Shape:", df.shape)

df.head()
```

```python
df.info()
```

```python
print(df.isnull().sum())
```

```python
print(df["category"].value_counts())
```

```python
import re

def clean_text(text):
    text = str(text).lower()

    # Remove URLs
    text = re.sub(r"http\S+|www\S+", " ", text)

    # Remove email addresses
    text = re.sub(r"\S+@\S+", " ", text)

    # Remove special characters and numbers
    text = re.sub(r"[^a-zA-Z\s]", " ", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

now subject add body:

```python
df["text"] = df["subject"].fillna("") + " " + df["body"].fillna("")
```



Cleaning:

```python
df["clean_text"] = df["text"].apply(clean_text)
```

Check:

```python
print(df[["text", "clean_text"]].head())
```

x and y prepare

```python
X = df["clean_text"]
y = df["category"]
```

```python
print(X.head())
print(y.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    stop_words="english",
    ngram_range=(1, 2),
    max_features=5000
)

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("Training data:", X_train_tfidf.shape)
print("Testing data:", X_test_tfidf.shape)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced"
)
```

```python
model.fit(X_train_tfidf, y_train)

print("Model training completed successfully!")
```

```python
y_pred = model.predict(X_test_tfidf)

print(y_pred)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", round(accuracy * 100, 2), "%")
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=model.classes_
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=model.classes_,
    yticklabels=model.classes_
)

plt.xlabel("Predicted Category")
plt.ylabel("Actual Category")
plt.title("Support Ticket Confusion Matrix")

plt.show()
```

```python
new_tickets = [
    {
        "subject": "Payment charged twice",
        "body": "I was charged two times for the same transaction."
    },
    {
        "subject": "Laptop VPN problem",
        "body": "I cannot connect to the company VPN from my laptop."
    },
    {
        "subject": "Leave approval pending",
        "body": "My leave request has not been approved yet."
    },
    {
        "subject": "Annual company event",
        "body": "Please share the date and location of the upcoming company event."
    },
    {
        "subject": "Refund still missing",
        "body": "My refund was approved but I have not received the money."
    }
]
```

```python
new_texts = []

for ticket in new_tickets:

    text = ticket["subject"] + " " + ticket["body"]

    cleaned_text = clean_text(text)

    new_texts.append(cleaned_text)

print(new_texts)
```

```python
new_tfidf = tfidf.transform(new_texts)

print("New ticket TF-IDF shape:", new_tfidf.shape)
```

```python
predictions = model.predict(new_tfidf)

for i in range(len(new_tickets)):

    print("Ticket", i + 1)
    print("Subject:", new_tickets[i]["subject"])
    print("Predicted Category:", predictions[i])
    print("-" * 50)
```

```python
probabilities = model.predict_proba(new_tfidf)

for i in range(len(new_tickets)):

    prediction = predictions[i]

    confidence = max(probabilities[i])

    print("Ticket:", i + 1)
    print("Subject:", new_tickets[i]["subject"])
    print("Category:", prediction)
    print("Confidence:", round(confidence * 100, 2), "%")
    print("-" * 50)
```

```python
def classify_ticket(subject, body):

    text = subject + " " + body

    text = clean_text(text)

    vector = tfidf.transform([text])

    probabilities = model.predict_proba(vector)[0]

    best_index = probabilities.argmax()

    category = model.classes_[best_index]

    confidence = probabilities[best_index]

    if confidence < 0.70:
        category = "Human Review"

    return category, confidence
```

```python
category, confidence = classify_ticket(
    "Need help",
    "Something is wrong and I need assistance."
)

print("Category:", category)
print("Confidence:", round(confidence * 100, 2), "%")
```
